```python
import pandas as pd

# Ruta al archivo limpio anterior
ruta = r"C:\Users\Vìctor\OneDrive\Desktop\ESP_Project\Datos\DATOS_ESP.xlsx"

# 1. Cargar hoja específica
df = pd.read_excel(ruta, sheet_name="SCHAO-473HS1UI")

# 2. Limpiar columnas
df.columns = (
    df.columns
    .str.strip()
    .str.upper()
    .str.replace(" +", " ", regex=True)
)

# 3. Detectar si la columna DATE está en formato numérico
df["DATE_ORIGINAL"] = df["DATE"]  # copia para revisar después

# Intentar convertir fechas: si es número, transformar como serial Excel
def convertir_fecha(x):
    try:
        if isinstance(x, (int, float)):
            return pd.to_datetime("1899-12-30") + pd.to_timedelta(x, unit="D")
        return pd.to_datetime(x, dayfirst=True, errors="coerce")
    except:
        return pd.NaT

df["DATE"] = df["DATE_ORIGINAL"].apply(convertir_fecha)

# 4. Mostrar cuántas fechas no se pudieron convertir
print("❗ Fechas inválidas:", df["DATE"].isna().sum())

# 5. Ordenar
df = df.sort_values("DATE").reset_index(drop=True)

# 6. Guardar limpio
df.to_excel("df_SCHAO473HS1UI_fechas_corregidas.xlsx", index=False)

# 7. Verificar visualmente
print(df[["DATE_ORIGINAL", "DATE"]].head(10))
```
