Importacion de librerias

```python
# --- 1. Librerías necesarias ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- 2. Cargar el archivo Excel ---
ruta = r'C:\Users\Vìctor\OneDrive\Desktop\ESP_Project\df_total_pozos_limpio.xlsx'
df = pd.read_excel(ruta)

# --- 3. Mostrar las primeras filas y dimensiones ---
print("Dimensiones:", df.shape)
df.head()
```

```python
df.tail(20)
```

Validar las columnas y tipos de datos

```python
# Ver columnas y tipos
df.info()

# Comprobar si alguna columna contiene "fecha"
fechas = [col for col in df.columns if 'fecha' in col.lower()]
print("Columnas que contienen fechas:", fechas)
```

Procesar y validar fechas

```python
# Supongamos que la columna se llama 'Fecha'
col_fecha = fechas[0]  # si solo hay una
df[col_fecha] = pd.to_datetime(df[col_fecha], errors='coerce')

# Rango de fechas
print("Fecha mínima:", df[col_fecha].min())
print("Fecha máxima:", df[col_fecha].max())
```

Verificar nulos y duplicados

```python
# Nulos por columna
df.isnull().sum()

# Duplicados por fecha (si aplica)
duplicados_fecha = df.duplicated(subset=[col_fecha])
print(f"Duplicados por fecha: {duplicados_fecha.sum()} registros")
```

```python
# Copia de seguridad del DataFrame
df_raw = df.copy()

# 1. Renombrar columnas automáticamente (sin tildes, espacios, símbolos raros)
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_', regex=False)
    .str.replace('%', 'pct', regex=False)
    .str.replace(r'[^\w\s]', '', regex=True)  # Elimina símbolos como ":"
)

# 2. Mostrar nombres resultantes
print("Columnas limpias:")
print(df.columns.tolist())
```

```python
# Verificar cuántos valores no nulos hay por columna unnamed
unnamed_cols = [col for col in df.columns if 'unnamed' in col]
df[unnamed_cols].notna().sum()
```

```python
df = df.drop(columns=[col for col in df.columns if 'unnamed' in col])
```

Verificar tipos de datos y valores nulos

```python
# Tipos de datos y resumen de valores nulos
df.info()

# Porcentaje de nulos por columna
(df.isnull().mean() * 100).round(2).sort_values(ascending=False)
```

Código completo para merge

```python
import pandas as pd

# 1. Cargar el dataset principal (si no lo tienes ya cargado)
# df = pd.read_excel("ruta_a_df_total_pozos_limpio.xlsx")  # si aplica

# 2. Cargar el archivo con mapeo POZO → CAMPO
ruta_campos = r'C:\Users\Vìctor\OneDrive\Desktop\ESP_Project\Datos\CampoSachamMejoresPozos.xlsx'
df_campo = pd.read_excel(ruta_campos, usecols=['POZO', 'CAMPO'])

# 3. Asegurarse que no haya duplicados (ya lo confirmamos)
df_mapeo = df_campo.drop_duplicates().copy()
df_mapeo.columns = ['pozo', 'campo_referencia']  # normalizar nombres

# 4. Merge solo por 'pozo'
df = df.merge(df_mapeo, on='pozo', how='left')

# 5. Rellenar campo si está vacío (NaN)
df['campo'] = df['campo'].fillna(df['campo_referencia'])

# 6. Eliminar columna auxiliar
df.drop(columns='campo_referencia', inplace=True)

# 7. Ver cuántos se llenaron
completados = df['campo'].notna().sum()
print(f"✅ Campo completado en {completados} registros de {len(df)}")
```
